# A sinusoid, from acquisition to eLabFTW

Run the cells in order after the one-time project setup. Data, code and the figure are saved locally in `data/` and uploaded in the background.

Author: Thibaut Jacqmin

```python
import numpy as np
import matplotlib.pyplot as plt
from elablink import AcquisitionAnalysisManager

aqm = AcquisitionAnalysisManager("data")
```

## Acquisition

Replace the generated signal with your instrument readings when ready.

```python
aqm.acquisition_cell("Sinusoid")
t = np.linspace(0, 1, 500)
y = np.sin(2 * np.pi * 5 * t)
aqm.save_acquisition(t=t, y=y)
```

## Analysis

Plot the saved data and keep the figure with the acquisition.

```python
aqm.analysis_cell()
plt.figure()
plt.plot(aqm.data.t, aqm.data.y)
plt.xlabel("Time (s)")
plt.ylabel("Signal (V)")
aqm.save_fig(name="sinusoid.png")
plt.show()
```

## Add notes to the eLabFTW experiment

Edits run in the background; reading the available choices waits for the server. The first available category and status are selected. This cell is optional.

```python
experiment = aqm.experiment
experiment.add_step("Sinusoid acquired and plotted.", checked=True)
experiment.add_comment("The signal is stable.")
experiment.set_extra_fields({"Frequency (Hz)": 5})
categories = list(experiment.categories)
statuses = list(experiment.statuses)
print("Categories:", categories)
print("Statuses:", statuses)
print("Tags:", experiment.tags)
if categories:
    experiment.category = categories[0]
if statuses:
    experiment.status = statuses[0]
```
